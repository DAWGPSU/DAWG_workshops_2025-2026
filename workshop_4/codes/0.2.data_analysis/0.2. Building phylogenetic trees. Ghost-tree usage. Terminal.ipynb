### Create a phylogenetic tree for 16S

```bash
# inputs
OTU_FILE='../../data/data_analysis/ps_microbiome.otu.csv'
TAXONOMY_FILE='../../data/data_analysis/ps_microbiome.taxonomy.csv'
# outputs
OTU_FILE_RENAMED='../../data/data_analysis/ghost_tree/ps_microbiome.otu.renamed.csv'
TAXONOMY_FILE_RENAMED='../../data/data_analysis/ghost_tree/ps_microbiome.taxonomy.renamed.csv'
TAXONOMY_FILE_GHOST='../../data/data_analysis/ghost_tree/ps_microbiome.taxonomy.ghost_tree.csv'
TAXONOMY_FASTA_FILE='../../data/data_analysis/ghost_tree/ps_microbiome.taxonomy.fasta'
TAXONOMY_ALIGNED_FASTA_FILE='../../data/data_analysis/ghost_tree/ps_microbiome.taxonomy_aligned.fasta'
TREE='../../data/data_analysis/ghost_tree/ps_microbiome.tree.nwk'
```

```bash
python scripts/convert_taxonomy_to_fasta.py \
    $OTU_FILE \
    $TAXONOMY_FILE \
    $OTU_FILE_RENAMED \
    $TAXONOMY_FILE_RENAMED \
    $TAXONOMY_FILE_GHOST \
    $TAXONOMY_FASTA_FILE 
```

```bash
muscle -in $TAXONOMY_FASTA_FILE \
       -out $TAXONOMY_ALIGNED_FASTA_FILE \
       -maxiters 2 -diags1
```

```bash
fasttree -nt -quiet $TAXONOMY_ALIGNED_FASTA_FILE > $TREE
```

### Prepare SILVA backbone data

```bash
# inputs
SILVA_DB_ALIGN="../../data/data_analysis/ghost_tree/SILVA_132_SSURef_Nr99_tax_silva_full_align_trunc.fasta.gz"
SILVA_TAX_MAP="../../data/data_analysis/ghost_tree/tax_slv_ssu_132.txt"
SILVA_TAX_ACC_TAXID="../../data/data_analysis/ghost_tree/tax_slv_ssu_132.acc_taxid"
# outputs
SILVA_DB_ALIGN_FASTA="../../data/data_analysis/ghost_tree/SILVA_132_SSURef_Nr99_tax_silva_full_align_trunc.fasta"
SILVA_FUNGI_ONLY_FILE="../../data/data_analysis/ghost_tree/silva_fungi_only.txt"
SILVA_FUNGI_ONLY_FILTERED_FILE="../../data/data_analysis/ghost_tree/silva_fungi_only_filtered.txt"
```

```bash
gunzip -k $SILVA_DB_ALIGN 
```

```bash
ghost-tree silva extract-fungi \
    $SILVA_DB_ALIGN_FASTA \
    $SILVA_TAX_ACC_TAXID \
    $SILVA_TAX_MAP \
    $SILVA_FUNGI_ONLY_FILE
 
```

```bash
ghost-tree filter-alignment-positions \
    $SILVA_FUNGI_ONLY_FILE \
    0.9 0.8 \
    $SILVA_FUNGI_ONLY_FILTERED_FILE 
```

### Building a Tree

```bash
# outputs
TAXONOMY_EXTENSIONS_FILE='../../data/data_analysis/ghost_tree/unite_db_tree_extensions.txt'
GHOST_TREE_OUT="../../data/data_analysis/ghost_tree/ghost_tree"
```

```bash
# Grouping the UNITE database tree extensions
ghost-tree extensions group-extensions \
    $TAXONOMY_FASTA_FILE \
    0.5 \
    $TAXONOMY_EXTENSIONS_FILE
```

```bash
# Building combining Silva backbone with UNITE tree extensions into a final tree
ghost-tree scaffold hybrid-tree-foundation-alignment \
    $TAXONOMY_EXTENSIONS_FILE \
    $TAXONOMY_FILE_GHOST  \
    $TAXONOMY_FASTA_FILE \
    $SILVA_FUNGI_ONLY_FILTERED_FILE \
    $GHOST_TREE_OUT
```
